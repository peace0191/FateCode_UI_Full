3.  **Colab에서 파일 복사:** 아래 예시 코드를 사용하여 Drive에 업로드된 이미지를 프로젝트 폴더로 복사합니다. Drive 경로와 파일 이름을 실제에 맞게 수정해야 합니다.


```python
# 예시: Google Drive에서 hanriver_premium 폴더로 이미지 복사
# '내 드라이브'에 'my_images'라는 폴더가 있고 그 안에 'image1.jpg'가 있다고 가정합니다.
# 원하는 Drive 경로와 파일 이름으로 수정해야 합니다.

import shutil
import os

drive_source_path_premium = '/content/drive/MyDrive/my_hanriver_premium_images/' # Drive 내의 한강뷰 이미지 폴더 경로
drive_source_path_normal = '/content/drive/MyDrive/my_normal_view_images/'     # Drive 내의 일반뷰 이미지 폴더 경로

dest_path_premium = '/content/CoreP_4910_DL_Project/data/hanriver_premium/'
dest_path_normal = '/content/CoreP_4910_DL_Project/data/normal_view/'

# 한강뷰 이미지 복사
if os.path.exists(drive_source_path_premium):
    for filename in os.listdir(drive_source_path_premium):
        if filename.lower().endswith(('.png', '.jpg', '.jpeg', '.gif')):
            shutil.copy(os.path.join(drive_source_path_premium, filename), dest_path_premium)
    print(f"{drive_source_path_premium} 의 이미지를 {dest_path_premium} 로 복사 완료.")
else:
    print(f"경로를 찾을 수 없습니다: {drive_source_path_premium}")

# 일반뷰 이미지 복사
if os.path.exists(drive_source_path_normal):
    for filename in os.listdir(drive_source_path_normal):
        if filename.lower().endswith(('.png', '.jpg', '.jpeg', '.gif')):
            shutil.copy(os.path.join(drive_source_path_normal, filename), dest_path_normal)
    print(f"{drive_source_path_normal} 의 이미지를 {dest_path_normal} 로 복사 완료.")
else:
    print(f"경로를 찾을 수 없습니다: {drive_source_path_normal}")

print("모든 이미지 복사 시도 완료.")
```

이미지를 업로드한 후에는 `!ls /content/CoreP_4910_DL_Project/data/hanriver_premium` 또는 `!ls /content/CoreP_4910_DL_Project/data/normal_view` 명령어를 사용하여 파일이 올바르게 위치했는지 확인할 수 있습니다.

```python
import torch
from torchvision.datasets import ImageFolder
from torch.utils.data import DataLoader
import torchvision.transforms as transforms

# DATA_DIR은 이전에 설정된 '/content/CoreP_4910_DL_Project/data/' 입니다.
DATA_DIR = '/content/CoreP_4910_DL_Project/data/'

# 이미지 전처리(변환) 정의 - 이전에 정의된 것과 동일
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# ImageFolder를 사용하여 데이터셋 로드 - 이전에 로드된 것과 동일
dataset = ImageFolder(root=DATA_DIR, transform=transform)

# 'normal_view' 클래스의 인덱스 찾기
normal_view_class_idx = dataset.class_to_idx['normal_view']
print(f"'normal_view' 클래스 인덱스: {normal_view_class_idx}")

# 'normal_view' 클래스에 속하는 이미지 필터링 및 개수 확인
normal_view_images_count = 0
for _, label in dataset.samples:
    if label == normal_view_class_idx:
        normal_view_images_count += 1

print(f"'normal_view' 폴더에서 로드된 이미지 총 개수: {normal_view_images_count}장")

# 예시: 'normal_view' 이미지 하나를 직접 로드 및 표시 (데이터셋에서 가져오는 방식)
# 실제 데이터셋에는 순서가 섞여 있을 수 있으므로, 인덱스를 직접 찾는 것은 복잡합니다.
# 여기서는 단순히 데이터셋에서 'normal_view'에 해당하는 첫 번째 이미지를 찾아 출력합니다.

first_normal_view_image = None
first_normal_view_label = None

for i in range(len(dataset)):
    image, label = dataset[i]
    if label == normal_view_class_idx:
        first_normal_view_image = image
        first_normal_view_label = label
        break

if first_normal_view_image is not None:
    print("\n'normal_view' 클래스의 첫 번째 이미지 (텐서):")
    print(first_normal_view_image.shape)
    print("해당 이미지의 레이블:", dataset.classes[first_normal_view_label])
else:
    print("\n'normal_view' 클래스의 이미지를 찾을 수 없습니다.")
```

```python
import os
import zipfile
import shutil

DATA_DIR = '/content/CoreP_4910_DL_Project/data/'
PREMIUM_DIR = os.path.join(DATA_DIR, 'hanriver_premium')
NORMAL_DIR = os.path.join(DATA_DIR, 'normal_view')

# Ensure base data directory and class subdirectories exist
os.makedirs(DATA_DIR, exist_ok=True)
os.makedirs(PREMIUM_DIR, exist_ok=True)
os.makedirs(NORMAL_DIR, exist_ok=True)

def process_directory(directory_path):
    print(f"Processing directory: {directory_path}")
    # The existence check is now less critical here since we ensured creation above
    if not os.path.exists(directory_path):
        # This line should ideally not be reached after os.makedirs
        print(f"Directory not found (unexpected): {directory_path}")
        return

    # Iterate through files in the directory
    for filename in os.listdir(directory_path):
        file_path = os.path.join(directory_path, filename)

        # If it's a zip file, extract it
        if filename.lower().endswith('.zip'):
            print(f"Unzipping {filename}...")
            try:
                with zipfile.ZipFile(file_path, 'r') as zip_ref:
                    zip_ref.extractall(directory_path)
                os.remove(file_path) # Remove the zip file after extraction
                print(f"Successfully unzipped and removed {filename}.")
            except zipfile.BadZipFile:
                print(f"Error: {filename} is a bad zip file.")
            except Exception as e:
                print(f"An error occurred while unzipping {filename}: {e}")
        # If it's an .ipynb file, remove it
        elif filename.lower().endswith('.ipynb'):
            print(f"Removing non-image file: {filename}")
            os.remove(file_path)
            print(f"Successfully removed {filename}.")


# Process both directories
process_directory(PREMIUM_DIR)
process_directory(NORMAL_DIR)

print("All zip files processed and non-image files removed from data directories.")

# Now, re-attempt to load images with ImageFolder
import torch
import torchvision.transforms as transforms
from torchvision.datasets import ImageFolder
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
import numpy as np

# 이미지 전처리(변환) 정의
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# ImageFolder를 사용하여 데이터셋 로드
# 이전에 에러가 났던 부분이지만, 이제 이미지가 있을 것으로 예상합니다.
try:
    dataset = ImageFolder(root=DATA_DIR, transform=transform)
    print(f"\n총 로드된 이미지 수: {len(dataset)}")
    print(f"클래스 목록: {dataset.classes}")

    if len(dataset) > 0:
        # DataLoader 생성
        batch_size = 4
        dataloader = DataLoader(dataset, batch_size=batch_size, shuffle=True)

        # 데이터 로더에서 한 배치(batch)의 이미지를 가져와 확인
        images, labels = next(iter(dataloader))

        print(f"한 배치 이미지 텐서 크기: {images.shape}")
        print(f"한 배치 레이블 텐서 크기: {labels.shape}")

        # 시각화를 위해 정규화를 되돌리는 함수
        def imshow(img):
            img = img / 2 + 0.5  # 비정규화
            npimg = img.numpy()
            plt.imshow(np.transpose(npimg, (1, 2, 0)))
            plt.show()

        # 첫 번째 배치 이미지 시각화 (예시)
        print("첫 번째 배치 이미지 (예시):")
        imshow(torchvision.utils.make_grid(images))
        print("해당 이미지들의 레이블:", [dataset.classes[label] for label in labels])
    else:
        print("데이터셋에 유효한 이미지가 없습니다. 폴더에 이미지가 올바르게 압축 해제되었는지 확인하세요.")

except FileNotFoundError as e:
    print(f"Error loading dataset: {e}. Please ensure image files are correctly placed in the class folders.")
except Exception as e:
    print(f"An unexpected error occurred during dataset loading or processing: {e}")
```

이미지 신뢰성 평가 (예: 이미지 품질, 일관성, 뷰가 한강인지 여부)

🧮 코드 샘플 (Python)

아래는 위 설계를 구현하기 위한 간단한 코드틀입니다. 이후 필요하시면 Streamlit 대시보드 버전까지 확장

```python
# library import
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import torch
from torchvision import transforms, models
from torch import nn
from PIL import Image
import glob
import os

# 1) 데이터 로드 (크롤러로 저장된 CSV 가정)
df = pd.read_csv('naver_listings_4910.csv', parse_dates=['date_listed'])
# 필터: E타입, 한강뷰 조건 등
df_filtered = df[(df['property_type']=='E타입') & (df['view']=='한강뷰')]

# 2) 최근 6개월 데이터 필터
cutoff = pd.Timestamp.today() - pd.DateOffset(months=6)
df_6m = df_filtered[df_filtered['date_listed'] >= cutoff]

# 3) 할인폭 계산: 기준가(현재 시세) 대비 10% 이상 저렴
current_price_estimate = 20_000 # 예: 억원 단위 변환 적용
df_discount = df_6m[df_6m['price'] <= current_price_estimate * 0.9]

# 4) 가격 트렌드 그래프
plt.figure(figsize=(10,6))
sns.lineplot(data=df_6m, x='date_listed', y='price', marker='o')
plt.axhline(current_price_estimate, color='red', linestyle='--', label='현재 추정가')
plt.title('최근 6개월 한강뷰 E타입 가격 추이')
plt.ylabel('가격 (억원)')
plt.legend()
plt.savefig('price_trend_6m.png')
plt.close()

# 5) 이미지 기반 신뢰성 평가 모델
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = models.resnet18(pretrained=True)
# 마지막 레이어 조정 (예: 안정성 판단 2클래스: 신뢰／주의)
num_ftrs = model.fc.in_features
model.fc = nn.Linear(num_ftrs, 2)
model = model.to(device)
model.eval()

transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225])
])

def evaluate_image(img_path):
    img = Image.open(img_path).convert('RGB')
    x = transform(img).unsqueeze(0).to(device)
    with torch.no_grad():
        logits = model(x)
        probs = torch.softmax(logits, dim=1).cpu().numpy()[0]
    return {'trustworthy': probs[0], 'questionable': probs[1]}

# 적용: 각 매물별 대표 이미지 평가
results = []
for idx, row in df_discount.iterrows():
    img_path = row['image_path']  # 이미지 경로 컬럼 가정
    res = evaluate_image(img_path)
    results.append({
        'listing_id': row['listing_id'],
        'price': row['price'],
        'trust_score': res['trustworthy']
    })

df_results = pd.DataFrame(results)
# 6) 신뢰도 상위 3건 추천
top3 = df_results.sort_values(by='trust_score', ascending=False).head(3)
print("추천 매물 Top 3:\n", top3)
```
