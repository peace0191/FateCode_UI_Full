# 1. 환경 설정 & 라이브러리 불러오기


```python
# CoreP / Colab 공통: 첫 셀
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, random_split
from torchvision import datasets, transforms
import matplotlib.pyplot as plt

device = "cuda" if torch.cuda.is_available() else "cpu"
print("사용 디바이스:", device)
```

2. 전처리 정의 + 데모 이미지 데이터셋(CIFAR-10) 불러오기

👉 여기가 “전처리 개념” 핵심:

Resize → 크기 맞추기

ToTensor → (H, W, C) 픽셀 → (C, H, W) 텐서

Normalize → 값 스케일 조정

```python
# 32x32 컬러 이미지 전처리 (CIFAR-10 기본 크기)
transform = transforms.Compose([
    transforms.ToTensor(),  # [0,255] -> [0,1] + 텐서
    transforms.Normalize(mean=[0.5, 0.5, 0.5],
                         std=[0.5, 0.5, 0.5])  # [-1,1] 범위로 정규화
])

# CIFAR-10 자동 다운로드 (train=True/False)
train_dataset_full = datasets.CIFAR10(
    root="./data", train=True, download=True, transform=transform
)
test_dataset = datasets.CIFAR10(
    root="./data", train=False, download=True, transform=transform
)

print("전체 train 샘플 수:", len(train_dataset_full))
print("test 샘플 수:", len(test_dataset))
```

# 3. 학습 / 검증 / 테스트 분리 (train/val/test split)

👉 “왜 나누는지”를 눈으로 보는 부분

train: 학습에 사용

val: 모델 튜닝용 (과적합 체크)

test: 최종 성적 확인용

```python
# 예시로 train 80% / val 20%
train_size = int(0.8 * len(train_dataset_full))
val_size = len(train_dataset_full) - train_size

train_dataset, val_dataset = random_split(
    train_dataset_full, [train_size, val_size]
)

print("train 샘플 수:", len(train_dataset))
print("val 샘플 수:", len(val_dataset))
```

# 4. DataLoader 만들기 (배치 단위로 묶어서 공급)

👉 “이미지가 자동으로 배치로 들어가는 통로” = DataLoader

```python
batch_size = 128

train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader   = DataLoader(val_dataset,   batch_size=batch_size, shuffle=False)
test_loader  = DataLoader(test_dataset,  batch_size=batch_size, shuffle=False)
```

#

# 5. 간단한 CNN 모델 정의 (3차 프로젝트 구조 축소판)

👉 CNN 구조 핵심

Conv → ReLU → MaxPool 반복으로 “특징(feature)” 추출

마지막에 Linear(fc)에서 클래스 분류

```python
class SimpleCNN(nn.Module):
    def __init__(self, num_classes=10):
        super().__init__()
        # 입력: 3채널(컬러 이미지)
        self.features = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, padding=1),  # 32x32 -> 32x32
            nn.ReLU(),
            nn.MaxPool2d(2),  # 32x32 -> 16x16

            nn.Conv2d(32, 64, kernel_size=3, padding=1), # 16x16 -> 16x16
            nn.ReLU(),
            nn.MaxPool2d(2),  # 16x16 -> 8x8

            nn.Conv2d(64, 128, kernel_size=3, padding=1), # 8x8 -> 8x8
            nn.ReLU(),
            nn.MaxPool2d(2)   # 8x8 -> 4x4
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128*4*4, 256),
            nn.ReLU(),
            nn.Dropout(0.5),  # 과적합 줄이기
            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x

model = SimpleCNN(num_classes=10).to(device)
print(model)
```

# 6. 손실함수 & 옵티마이저 설정

👉 손실(loss) = 예측과 정답의 차이

이미지 분류 → 보통 CrossEntropyLoss

최적화 → Adam / SGD

```python
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=1e-3)
```

# 7. 학습/검증 루프 (과적합/과소적합 보는 부분)

👉 여기서 에포크마다 train / val loss, accuracy 를 출력해서
“과적합인지, 부족하게 학습했는지” 감을 잡을 수 있습니다.

```python
def train_one_epoch(model, loader, criterion, optimizer, device):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0

    for images, labels in loader:
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()          # 역전파
        optimizer.step()          # 파라미터 업데이트

        running_loss += loss.item() * images.size(0)
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    epoch_loss = running_loss / total
    epoch_acc = correct / total
    return epoch_loss, epoch_acc


def evaluate(model, loader, criterion, device):
    model.eval()
    running_loss = 0.0
    correct = 0
    total = 0

    with torch.no_grad():
        for images, labels in loader:
            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)

            running_loss += loss.item() * images.size(0)
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

    epoch_loss = running_loss / total
    epoch_acc = correct / total
    return epoch_loss, epoch_acc
```

# 7-1. 실제로 학습 돌리기

```python
num_epochs = 5  # 처음엔 5로 시작해 보고, 익숙해지면 10~20까지 늘려보세요.

train_losses, val_losses = [], []
train_accs, val_accs = [], []

for epoch in range(1, num_epochs+1):
    train_loss, train_acc = train_one_epoch(model, train_loader, criterion, optimizer, device)
    val_loss, val_acc     = evaluate(model, val_loader, criterion, device)

    train_losses.append(train_loss)
    val_losses.append(val_loss)
    train_accs.append(train_acc)
    val_accs.append(val_acc)

    print(f"[Epoch {epoch}/{num_epochs}] "
          f"train_loss={train_loss:.4f} val_loss={val_loss:.4f} "
          f"train_acc={train_acc*100:.2f}% val_acc={val_acc*100:.2f}%")
```

# 8. 학습 곡선 그려서 과적합/과소적합 느낌 잡기

```python
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.plot(train_losses, label="train_loss")
plt.plot(val_losses,   label="val_loss")
plt.legend()
plt.title("Loss Curve")

plt.subplot(1,2,2)
plt.plot(train_accs, label="train_acc")
plt.plot(val_accs,   label="val_acc")
plt.legend()
plt.title("Accuracy Curve")

plt.show()
```

train_loss ↓, val_loss도 같이 ↓ → 좋은 학습

train_loss ↓, val_loss ↑ → 과적합(훈련 데이터에만 너무 잘 맞음)

둘 다 높고 변화 적음 → 과소적합(모델이 너무 약하거나 학습이 부족)

# 9. 테스트 데이터에서 최종 성능 확인

```python
test_loss, test_acc = evaluate(model, test_loader, criterion, device)
print(f"최종 Test Loss: {test_loss:.4f}, Test Accuracy: {test_acc*100:.2f}%")
```

## 2D 이미지 데이터 로딩 및 Tensor 변환 (기존 CIFAR-10 예시)

이전 셀 `WqcP86rjkdjH`에서 다음과 같은 과정을 통해 2D 이미지 데이터가 로드되고 Tensor로 변환됩니다.

1.  **`transforms.Compose`**: 여러 전처리 단계를 하나로 묶습니다.
2.  **`transforms.ToTensor()`**: PIL Image나 NumPy 배열 형태의 이미지를 PyTorch Tensor로 변환합니다. 이 과정에서 픽셀 값은 `[0, 255]` 범위에서 `[0, 1]` 범위로 자동 스케일링됩니다. 또한 이미지의 채널 순서가 `(H, W, C)`에서 `(C, H, W)`로 변경됩니다.
3.  **`transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5])`**: `[0, 1]` 범위의 텐서 픽셀 값을 `[-1, 1]` 범위로 정규화합니다. 이는 딥러닝 모델의 학습 안정성을 높이는 데 도움을 줍니다.
4.  **`datasets.CIFAR10(...)`**: `root` 경로에 CIFAR-10 데이터셋을 다운로드하고, 정의된 `transform`을 적용하여 `Dataset` 객체를 생성합니다.

## 3D 이미지 데이터 로딩 및 Tensor 변환 (합성 데이터 예시)

3D 이미지의 경우, 일반적으로 `nn.Conv3d`를 사용하기 위해 입력 텐서의 형태가 `(N, C, D, H, W)`여야 합니다. 여기서 `N`은 배치 크기, `C`는 채널 수, `D`는 깊이(Depth), `H`는 높이(Height), `W`는 너비(Width)를 나타냅니다.

다음은 임의의 합성 3D 데이터를 생성하고 이를 PyTorch Tensor로 변환하며 `DataLoader`를 구성하는 예시 코드입니다. 실제 3D 의료 영상(CT, MRI 등)의 경우 데이터를 읽어와서 이와 유사한 형태로 전처리하게 됩니다.

```python
import torch
from torch.utils.data import Dataset, DataLoader

# 1. 합성 3D 이미지 데이터셋 정의
class Synthetic3DDataset(Dataset):
    def __init__(self, num_samples=100, depth=16, height=32, width=32, num_channels=1, num_classes=10, transform=None):
        self.num_samples = num_samples
        self.depth = depth
        self.height = height
        self.width = width
        self.num_channels = num_channels
        self.num_classes = num_classes
        self.transform = transform

        # 임의의 3D 이미지와 라벨 생성
        self.data = torch.randn(num_samples, num_channels, depth, height, width)
        self.labels = torch.randint(0, num_classes, (num_samples,))

    def __len__(self):
        return self.num_samples

    def __getitem__(self, idx):
        image = self.data[idx]
        label = self.labels[idx]

        if self.transform:
            image = self.transform(image)
        return image, label

# 2. 3D 이미지 전처리 (Tensor로 변환 + 정규화)
# 3D 텐서에 직접 적용할 수 있는 Normalize 함수가 torchvision.transforms에 직접적으로 없으므로,
# Custom Transform을 정의하거나, ToTensor()로 범위 조정 후 수동으로 정규화를 적용합니다.
# 여기서는 임의의 범위 [-1, 1]로 맞추는 예시를 듭니다.

# ToTensor()는 이미지를 [0,1]로 스케일링하며 (H,W,C) -> (C,H,W) 변환을 주로 합니다.
# 3D 데이터(D,H,W)나 (C,D,H,W) 형태의 텐서에 대해서는 ToTensor()가 직접적으로
# [0,1] 스케일링을 해주지 않을 수 있으므로, 직접 스케일링을 적용하는 것이 좋습니다.

# Custom Transform for 3D data
class Normalize3D:
    def __init__(self, mean, std):
        self.mean = torch.tensor(mean).view(1, -1, 1, 1, 1) if isinstance(mean, (list, tuple)) else torch.tensor(mean)
        self.std = torch.tensor(std).view(1, -1, 1, 1, 1) if isinstance(std, (list, tuple)) else torch.tensor(std)

    def __call__(self, tensor):
        # 입력 텐서의 채널이 mean/std와 맞는지 확인
        if tensor.dim() == 4 and self.mean.dim() == 1: # C, D, H, W (1채널) -> mean/std는 1개
            return (tensor - self.mean.view(1,-1,1,1)) / self.std.view(1,-1,1,1)
        elif tensor.dim() == 5 and self.mean.dim() > 1: # N, C, D, H, W (다채널) -> mean/std는 C개
            return (tensor - self.mean) / self.std
        elif tensor.dim() == 4 and self.mean.dim() > 1: # C, D, H, W (다채널) -> mean/std는 C개
            return (tensor - self.mean.view(-1, 1, 1, 1)) / self.std.view(-1, 1, 1, 1)
        else:
            return (tensor - self.mean) / self.std


# 3D 이미지용 변환 정의 (합성 데이터는 이미 Tensor 형태이므로 ToTensor는 필요 없음)
# 만약 NumPy 배열이나 다른 형식이라면 ToTensor()와 유사한 직접 변환 로직이 필요함
transform_3d = transforms.Compose([
    Normalize3D(mean=[0.5], std=[0.5]) # 1채널 3D 이미지용 정규화
])

# 3. 3D 합성 데이터셋 인스턴스 생성
synthetic_3d_dataset = Synthetic3DDataset(num_samples=1000, depth=16, height=32, width=32, num_channels=1, num_classes=10, transform=transform_3d)

print(f"합성 3D 데이터셋 샘플 수: {len(synthetic_3d_dataset)}")
print(f"첫 번째 이미지 텐서 shape: {synthetic_3d_dataset[0][0].shape}")
print(f"첫 번째 이미지 텐서 값 범위: {synthetic_3d_dataset[0][0].min():.2f} ~ {synthetic_3d_dataset[0][0].max():.2f}")

# 4. DataLoader 구성
batch_size_3d = 8
three_d_loader = DataLoader(synthetic_3d_dataset, batch_size=batch_size_3d, shuffle=True)

# DataLoader에서 첫 번째 배치 확인
first_batch_images, first_batch_labels = next(iter(three_d_loader))

print(f"\nDataLoader 첫 번째 배치 이미지 shape: {first_batch_images.shape}")
print(f"DataLoader 첫 번째 배치 라벨 shape: {first_batch_labels.shape}")
```

## 3D CNN 모델 정의 (Simple3DCNN)

2D CNN 모델(`SimpleCNN`)과 유사하게, 3D 이미지 데이터를 처리하기 위한 `Simple3DCNN` 모델을 정의합니다.

주요 변경 사항:

*   **`nn.Conv2d` → `nn.Conv3d`**: 2D 컨볼루션 레이어를 3D 컨볼루션 레이어로 변경합니다. 입력 텐서의 형태는 `(N, C, D, H, W)`가 됩니다.
*   **`nn.MaxPool2d` → `nn.MaxPool3d`**: 2D 맥스 풀링 레이어를 3D 맥스 풀링 레이어로 변경합니다.
*   **`nn.Linear` 입력 크기 조정**: 컨볼루션 및 풀링 과정을 거치면서 텐서의 깊이, 높이, 너비가 모두 줄어들기 때문에, 마지막 `Flatten()` 이후 `Linear` 레이어의 입력 크기를 `128 * 2 * 4 * 4` (여기서는 예시 깊이 16이 2로 줄어들었을 경우)와 같이 새로운 텐서 크기에 맞춰 조정해야 합니다. 아래 예시에서는 `depth=16`이 3번의 MaxPool3d(kernel_size=2)를 거쳐 `16 -> 8 -> 4 -> 2`로 줄어든 것을 반영했습니다.

```python
class Simple3DCNN(nn.Module):
    def __init__(self, num_classes=10, in_channels=1):
        super().__init__()
        # 입력: in_channels (합성 데이터는 1채널)
        self.features = nn.Sequential(
            nn.Conv3d(in_channels, 32, kernel_size=3, padding=1),  # D,H,W 유지 (ex: 16x32x32 -> 16x32x32)
            nn.ReLU(),
            nn.MaxPool3d(2),  # D,H,W 절반으로 줄어듬 (ex: 16x32x32 -> 8x16x16)

            nn.Conv3d(32, 64, kernel_size=3, padding=1), # D,H,W 유지 (ex: 8x16x16 -> 8x16x16)
            nn.ReLU(),
            nn.MaxPool3d(2),  # D,H,W 절반으로 줄어듬 (ex: 8x16x16 -> 4x8x8)

            nn.Conv3d(64, 128, kernel_size=3, padding=1), # D,H,W 유지 (ex: 4x8x8 -> 4x8x8)
            nn.ReLU(),
            nn.MaxPool3d(2)   # D,H,W 절반으로 줄어듬 (ex: 4x8x8 -> 2x4x4)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            # Flatten된 피처 맵의 크기 계산: 채널 * 깊이 * 높이 * 너비
            # 예시: 128 * 2 * 4 * 4 = 4096
            nn.Linear(128 * 2 * 4 * 4, 256), # Output from last MaxPool3d (depth=2, height=4, width=4)
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x

# 모델 인스턴스 생성 (합성 데이터는 1채널이므로 in_channels=1)
model_3d = Simple3DCNN(num_classes=10, in_channels=1).to(device)
print(model_3d)

# 임의의 3D 텐서를 모델에 통과시켜 출력 shape 확인
# 배치 크기 8, 채널 1, 깊이 16, 높이 32, 너비 32
input_3d_tensor = torch.randn(8, 1, 16, 32, 32).to(device)
output_3d = model_3d(input_3d_tensor)
print(f"\n모델 출력 shape: {output_3d.shape}")
```

```python
(이미지/텍스트) → 전처리 → Dataset → DataLoader → 모델 → 손실/옵티마이저
→ 학습 반복 → 체크포인트 저장 → 추론(서비스)
```

여기서는 이전에 생성한 `model_3d`와 `three_d_loader`를 사용합니다.

# 🔵 3D CNN 데모 코드 (데이터 없이도 실행됨 — synthetic data)

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader

# --------------------------
# 1) Synthetic 3D Dataset
# --------------------------
class Demo3DDataset(Dataset):
    def __init__(self, num_samples=40):
        super().__init__()
        self.num_samples = num_samples
        # 1채널, Depth=16, H=64, W=64
        self.data = torch.randn(num_samples, 1, 16, 64, 64)
        self.labels = torch.randint(0, 2, (num_samples,))

    def __len__(self):
        return self.num_samples

    def __getitem__(self, idx):
        return self.data[idx], self.labels[idx]

train_loader = DataLoader(Demo3DDataset(), batch_size=4, shuffle=True)

# --------------------------
# 2) 3D CNN 모델 정의
# --------------------------
class Simple3DCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.model = nn.Sequential(
            nn.Conv3d(1, 8, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool3d(2),
            nn.Conv3d(8, 16, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool3d(2),
        )
        # 출력은 (Batch, 16, 4, 16, 16)
        self.fc = nn.Linear(16 * 4 * 16 * 16, 2)

    def forward(self, x):
        x = self.model(x)
        x = x.view(x.size(0), -1)
        return self.fc(x)

model = Simple3DCNN()
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# --------------------------
# 3) 학습 루프
# --------------------------
for epoch in range(3):
    for x, y in train_loader:
        optimizer.zero_grad()
        y_pred = model(x)
        loss = criterion(y_pred, y)
        loss.backward()
        optimizer.step()
    print(f"Epoch {epoch+1} | Loss = {loss.item():.4f}")

print("🔥 3D CNN Demo Training Completed!")
```

Conv3D → ReLU → Pooling → Flatten → FC 구조

🔵 ImageFolder + ResNet18 학습 코드 (3D 아님 / 4910 프로젝트용)

```python
import torch
import torch.nn as nn
from torchvision import datasets, transforms, models
from torch.utils.data import DataLoader

# --------------------------
# 1) 이미지 전처리 (preprocessing)
# --------------------------
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

train_dir = "/content/data/train"
val_dir = "/content/data/val"

train_set = datasets.ImageFolder(train_dir, transform=transform)
val_set = datasets.ImageFolder(val_dir, transform=transform)

train_loader = DataLoader(train_set, batch_size=16, shuffle=True)
val_loader = DataLoader(val_set, batch_size=16, shuffle=False)

# 클래스 이름 확인
print("Class index -> name :", train_set.class_to_idx)

# --------------------------
# 2) ResNet18 모델 정의
# --------------------------
model = models.resnet18(weights="IMAGENET1K_V1")
model.fc = nn.Linear(model.fc.in_features, 2)
model = model.to("cuda" if torch.cuda.is_available() else "cpu")

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-4)

# --------------------------
# 3) 학습 함수
# --------------------------
device = "cuda" if torch.cuda.is_available() else "cpu"

def train_one_epoch(epoch):
    model.train()
    total_loss = 0
    for x, y in train_loader:
        x, y = x.to(device), y.to(device)
        optimizer.zero_grad()
        pred = model(x)
        loss = criterion(pred, y)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    print(f"[Epoch {epoch}] Train Loss: {total_loss/len(train_loader):.4f}")

def validate():
    model.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for x, y in val_loader:
            x, y = x.to(device), y.to(device)
            pred = model(x)
            _, predicted = torch.max(pred, 1)
            correct += (predicted == y).sum().item()
            total += y.size(0)
    print(f"Validation Acc: {correct/total:.4f}")

# --------------------------
# 4) 실행
# --------------------------
for epoch in range(5):
    train_one_epoch(epoch+1)
    validate()

torch.save(model.state_dict(), "signiel4910_resnet18.pth")
print("🔥 Training Completed — Model Saved!")
```

⭐ 이 코드가 “3차 프로젝트 실패 원인 5가지”를 모두 해결합니다

1. 전처리 개념 없음	transform.Resize, Normalize 로 명확히 처리
2. Train/Val split 개념 없음	ImageFolder + train/val 분리
3. CNN 구조 이해 부족	ResNet18(특징추출) → FC(분류) 흐름 명확
4. Overfitting 개념 미약	val accuracy 확인 코드 포함
5. 전체 파이프라인 부재	“불러오기 → 전처리 → 모델 → 학습 → 저장 → 추론” 전체 구조 제공
