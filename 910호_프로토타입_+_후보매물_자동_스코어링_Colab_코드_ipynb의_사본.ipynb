## 1) 4910호 프로토타입 + 후보매물 자동 스코어링 Colab 코드

```python
# =========================
# 1. 기본 세팅 & 압축 풀기
# =========================
import os, glob, shutil
from pathlib import Path

BASE_DIR = "/content"
ZIP_PATH = "/content/all_4910_assets.zip"  # 업로드한 zip 경로

# 압축 풀 폴더
ASSET_DIR = os.path.join(BASE_DIR, "4910_assets")
os.makedirs(ASSET_DIR, exist_ok=True)

# 압축 해제
!unzip -q "{ZIP_PATH}" -d "{ASSET_DIR}"

print("압축 해제 완료:", ASSET_DIR)
print("파일 개수:", len(os.listdir(ASSET_DIR)))
```

```python
# ====================================
# 2. 폴더 구조 자동 생성 + 파일 분류
#    - refs_4910E  : 4910호 실사진(참고 이미지)
#    - stock_4910E : 책자/홍보컷(E타입 샘플, 외경 등)
#    - candidates_4910E : 네이버 확인매물 등 후보매물(나중에 직접 업로드)
# ====================================
DATA_ROOT = os.path.join(BASE_DIR, "data_4910E")
REF_DIR   = os.path.join(DATA_ROOT, "refs_4910E")
STOCK_DIR = os.path.join(DATA_ROOT, "stock_4910E")
CAND_DIR  = os.path.join(DATA_ROOT, "candidates_4910E")

os.makedirs(REF_DIR, exist_ok=True)
os.makedirs(STOCK_DIR, exist_ok=True)
os.makedirs(CAND_DIR, exist_ok=True)

jpg_files = glob.glob(os.path.join(ASSET_DIR, "*.jpg"))

for src in jpg_files:
    name = os.path.basename(src)
    # 4910호 실사진 → 프로토타입용
    if name.startswith("4910호"):
        dst = os.path.join(REF_DIR, name)
    # E타입, 외경/홍보이미지 → 스톡/홍보컷
    elif name.startswith("E타입") or "홍보이미지" in name or "외경" in name:
        dst = os.path.join(STOCK_DIR, name)
    else:
        # 일단 나머지는 참고용으로 stock에 넣어두기
        dst = os.path.join(STOCK_DIR, name)
    shutil.copy(src, dst)

print("참고(4910호) 이미지 수:", len(os.listdir(REF_DIR)))
print("스톡/홍보 이미지 수:", len(os.listdir(STOCK_DIR)))
print("후보매물 폴더(직접 업로드 필요) 경로:", CAND_DIR)
```

## 3. 이미지 임베딩 모델(ResNet50) 로딩 + 임베딩 함수


```python
# ===============================
# 3. 이미지 임베딩 모델 세팅
#    - ResNet50의 마지막 직전 레이어를 특징벡터로 사용
# ===============================
import torch
import torch.nn as nn
from PIL import Image
import torchvision.models as models
import torchvision.transforms as T

device = "cuda" if torch.cuda.is_available() else "cpu"
print("사용 디바이스:", device)

# ResNet50 불러오기
resnet = models.resnet50(pretrained=True)
resnet.eval()

# 마지막 FC층 제거 → 2048차원 특징벡터
feature_extractor = nn.Sequential(*list(resnet.children())[:-1]).to(device)
for p in feature_extractor.parameters():
    p.requires_grad = False

# 이미지 전처리 (ImageNet 기준)
transform = T.Compose([
    T.Resize(256),
    T.CenterCrop(224),
    T.ToTensor(),
    T.Normalize(
        mean=[0.485, 0.456, 0.406],
        std =[0.229, 0.224, 0.225]
    ),
])

def load_image(path):
    img = Image.open(path).convert("RGB")
    return transform(img).unsqueeze(0).to(device)

def embed_image(path):
    """단일 이미지 → 2048차원 L2 정규화 벡터"""
    x = load_image(path)
    with torch.no_grad():
        feat = feature_extractor(x)  # (1, 2048, 1, 1)
    feat = feat.view(2048)
    feat = feat / feat.norm(p=2)
    return feat

def embed_folder(folder):
    paths = sorted(glob.glob(os.path.join(folder, "*.jpg")))
    embeds = []
    for p in paths:
        try:
            emb = embed_image(p)
            embeds.append((p, emb))
        except Exception as e:
            print("임베딩 실패:", p, e)
    return embeds
```

## 4. 4910호 프로토타입 + 스톡(홍보컷) 프로토타입 생성|

```python
# ======================================
# 4. 4910호 실사진 프로토타입 생성
# ======================================
ref_embeds = embed_folder(REF_DIR)
assert len(ref_embeds) > 0, "refs_4910E 폴더에 4910호 실사진이 필요합니다."

ref_paths, ref_vecs = zip(*ref_embeds)
ref_vecs = torch.stack(ref_vecs, dim=0)  # (N, 2048)

proto_4910 = ref_vecs.mean(0)
proto_4910 = proto_4910 / proto_4910.norm(p=2)

print("4910 프로토타입 벡터 차원:", proto_4910.shape)
print("참고 이미지 수:", len(ref_paths))
```

```python
# ======================================
# 5. 스톡/홍보컷 프로토타입 (가짜/과장 인테리어 비교용)
# ======================================
stock_embeds = embed_folder(STOCK_DIR)
if len(stock_embeds) > 0:
    stock_paths, stock_vecs = zip(*stock_embeds)
    stock_vecs = torch.stack(stock_vecs, dim=0)
    proto_stock = stock_vecs.mean(0)
    proto_stock = proto_stock / proto_stock.norm(p=2)
    has_stock_proto = True
    print("스톡/홍보컷 프로토타입 생성 완료, 이미지 수:", len(stock_paths))
else:
    proto_stock = None
    has_stock_proto = False
    print("⚠ stock_4910E 폴더가 비어 있어 스톡 프로토타입을 만들 수 없습니다.")
```

## 5. 후보 매물 임베딩 + 유사도 / 진위 / 프리미엄 스코어 계산

```python
# ===============================
# 6. 후보 매물(candidates_4910E) 임베딩
#    - 여기 폴더에 네이버/유튜브/직접촬영 후보매물 사진 넣어두면 자동 분석
# ===============================
cand_embeds = embed_folder(CAND_DIR)
if len(cand_embeds) == 0:
    print("⚠ candidates_4910E 폴더에 후보 매물 사진을 업로드해 주세요.")
else:
    print("후보 매물 이미지 수:", len(cand_embeds))
```

```python
# ===============================
# 7. 유사도 + 가짜/스톡 의심 점수 + 프리미엄 점수 계산
#    - similarity_4910 : 4910호 프로토타입과의 코사인 유사도
#    - similarity_stock : 스톡 프로토타입과의 유사도
#    - real_minus_stock : 실사진 유사도 - 스톡 유사도 (낮으면 가짜/보정 의심)
#    - premium_score : (옵션) 향후 프리미엄 요소 프로토타입 추가시 확장
# ===============================
import pandas as pd

results = []

for path, emb in cand_embeds:
    sim_4910 = float(torch.dot(emb, proto_4910).cpu())
    if has_stock_proto:
        sim_stock = float(torch.dot(emb, proto_stock).cpu())
        real_minus_stock = sim_4910 - sim_stock
    else:
        sim_stock = None
        real_minus_stock = None

    # 프리미엄 점수 (지금은 placeholder: 일단 4910 유사도 자체를 프리미엄으로 사용)
    # 나중에 'premium_view_4910E', 'premium_kitchen_4910E' 같은 폴더 만들어
    # 각각 프로토타입을 만들면 가중합으로 교체 가능
    premium_score = sim_4910

    results.append({
        "filename": os.path.basename(path),
        "path": path,
        "similarity_4910": sim_4910,
        "similarity_stock": sim_stock,
        "real_minus_stock": real_minus_stock,
        "premium_score": premium_score,
    })

df = pd.DataFrame(results)
df_sorted = df.sort_values("similarity_4910", ascending=False).reset_index(drop=True)

print("후보 매물 스코어링 결과 (상위 5개):")
df_sorted.head()
```

```python
# 가짜/홍보컷 의심 플래그 추가 (임계값은 나중에 조정 가능)
FAKE_THRESHOLD = -0.03  # 예시값: 실사진보다 스톡에 더 가까운 편이면 의심

if has_stock_proto:
    df_sorted["maybe_fake_or_stock"] = df_sorted["real_minus_stock"] < FAKE_THRESHOLD
else:
    df_sorted["maybe_fake_or_stock"] = None

df_sorted.head(10)
```

```python
# ===============================
# 10. PDF 리포트 자동 생성
#     - 1페이지: 요약 텍스트
#     - 2페이지: 테이블(Top10)
#     - 3페이지: 유사도 막대 그래프
# ===============================
from matplotlib.backends.backend_pdf import PdfPages
import numpy as np
import datetime

REPORT_PATH = "/content/4910E_similarity_report.pdf"

with PdfPages(REPORT_PATH) as pdf:
    # Page 1: Summary Text
    fig1 = plt.figure(figsize=(8.27, 11.69))  # A4 세로
    plt.axis("off")
    now = datetime.datetime.now().strftime("%Y-%m-%d %H:%M")

    summary_lines = [
        "롯데월드타워몰 시그니엘레지던스 4910호 E타입",
        "이미지 프로토타입 기반 유사 매물 분석 리포트",
        "",
        f"생성 일시: {now}",
        "",
        f"참고(4910호) 이미지 수: {len(ref_paths)}장",
        f"스톡/홍보 이미지 수: {len(stock_embeds) if has_stock_proto else 0}장",
        f"후보 매물 이미지 수: {len(cand_embeds)}장",
        "",
        "지표 설명:",
        "- similarity_4910 : 4910호 실사진 프로토타입과의 코사인 유사도",
        "- similarity_stock : 스톡/홍보컷 프로토타입과의 유사도",
        "- real_minus_stock : 실사진 유사도 - 스톡 유사도",
        "   (음수에 가까울수록 카다로그/보정/스톡 사진 가능성 ↑)",
        "- premium_score : 현재는 similarity_4910을 그대로 사용",
    ]
    text = "\n".join(summary_lines)
    plt.text(0.02, 0.98, text, va="top", fontsize=11)
    pdf.savefig(fig1)
    plt.close(fig1)

    # Page 2: Table (Top10)
    fig2 = plt.figure(figsize=(11.69, 8.27))  # A4 가로
    plt.axis("off")

    cols = ["rank", "filename", "similarity_4910", "similarity_stock", "real_minus_stock", "maybe_fake_or_stock"]
    table_data = []
    for i, row in top10.reset_index(drop=True).iterrows():
        table_data.append([
            i + 1,
            row["filename"],
            f"{row['similarity_4910']:.3f}",
            "" if row["similarity_stock"] is None else f"{row['similarity_stock']:.3f}",
            "" if row["real_minus_stock"] is None else f"{row['real_minus_stock']:.3f}",
            "" if row["maybe_fake_or_stock"] is None else ("의심" if row["maybe_fake_or_stock"] else ""),
        ])

    table = plt.table(
        cellText=table_data,
        colLabels=["Rank", "파일명", "유사도(4910)", "유사도(스톡)", "실사진-스톡", "가짜/스톡 의심"],
        loc="center",
        cellLoc="center"
    )
    table.auto_set_font_size(False)
    table.set_fontsize(8)
    table.scale(1, 1.5)

    plt.title("Top 10 Similar Listings to 4910E", pad=20)
    pdf.savefig(fig2)
    plt.close(fig2)

    # Page 3: Bar Chart (유사도)
    fig3 = plt.figure(figsize=(11.69, 8.27))
    plt.bar(top10["filename"], top10["similarity_4910"])
    plt.xticks(rotation=70, ha="right")
    plt.ylabel("Similarity to 4910 Prototype")
    plt.title("Top 10 Similar Listings to 4910E (Image-based)")
    plt.tight_layout()
    pdf.savefig(fig3)
    plt.close(fig3)

REPORT_PATH
```
